```python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters of the Gaussian mixture model
mu = np.array([-2, 1, 4])
sigma = np.array([np.sqrt(1 / 2), np.sqrt(2), 1])
weights = np.array([0.5, 0.2, 0.3])

# Define the range of x values
x = np.arange(-10, 10, 0.05)

# Define the manual implementation of the Gaussian PDF
def gaussian_pdf(x, mu, sigma):
    return (1 / (np.sqrt(2 * np.pi) * sigma)) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

# Calculate the PDF for each component and sum them up
pdf = np.zeros_like(x)
for i in range(len(mu)):
    pdf += weights[i] * gaussian_pdf(x, mu[i], sigma[i])

# Plot the PDF
plt.figure(figsize=(8, 6))
plt.plot(x, pdf, label='Gaussian Mixture Model')
plt.title('Probability Density Function of Gaussian Mixture Model')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters of the Gaussian mixture model
mean_values = np.array([-2, 1, 4])
standard_deviations = np.array([np.sqrt(1 / 2), np.sqrt(2), 1])
weights = np.array([0.5, 0.2, 0.3])

# Define the range of x values
x_values = np.arange(-10, 10, 0.05)

# Define the manual implementation of the Gaussian PDF
def gaussian_probability_density(x, mu, sigma):
    return (1 / (np.sqrt(2 * np.pi) * sigma)) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

# Calculate the PDF for each component and sum them up
pdf_values = np.zeros_like(x_values)
for i in range(len(mean_values)):
    pdf_values += weights[i] * gaussian_probability_density(x_values, mean_values[i], standard_deviations[i])

# Plot the PDF
plt.figure(figsize=(8, 6))
plt.plot(x_values, pdf_values, label='Gaussian Mixture Model')
plt.title('Probability Density Function of Gaussian Mixture Model')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Q2a)
import matplotlib.pyplot as plt

# Define the data points
data = {
    'x1': [76, 77, 76, 78, 77, 78],
    'x2': [76, 76, 77, 77, 78, 78],
    'y': [-1, -1, -1, 1, 1, 1]
}

# Plotting the data
plt.figure(figsize=(6, 6))
plt.scatter(data['x1'], data['x2'], c=data['y'], cmap='coolwarm', s=100)
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Plot of Data Points')
plt.grid(True)
plt.show()

# Q2c)
import numpy as np

class LinearSVC:
    def __init__(self, learning_rate=0.0001, epochs=1000, verbose=False):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.verbose = verbose

    def fit(self, X, y):
        # Initialize weights and bias
        self.weights = np.zeros(X.shape[1])
        self.bias = 0

        # Initialize Lagrange multipliers
        alpha = np.zeros(X.shape[0])

        for epoch in range(1, self.epochs + 1):
            # Compute predictions
            predictions = np.dot(X, self.weights) + self.bias

            # Compute hinge loss
            loss = 1 - y * predictions
            loss[loss < 0] = 0

            # Compute gradient of weights and bias
            weights_gradient = -2 * np.dot(X.T, loss * y)
            bias_gradient = -2 * np.sum(loss * y)

            # Update weights and bias
            self.weights -= self.learning_rate * weights_gradient
            self.bias -= self.learning_rate * bias_gradient

            # Update Lagrange multipliers for support vectors
            alpha[loss > 0] += 1

            if self.verbose and epoch % 100 == 0:
                print(f'Epoch {epoch}/{self.epochs}, Loss: {np.mean(loss)}')

        # Record support vectors
        self.support_vectors_ = X[alpha > 0]

# Define the training data
X_train = np.array([[76, 76], [77, 76], [76, 77], [78, 77], [77, 78], [78, 78]])
y_train = np.array([-1, -1, -1, 1, 1, 1])

# Create and train the model
svm = LinearSVC(learning_rate=0.0001, epochs=1000, verbose=True)
svm.fit(X_train, y_train)

# Get support vectors
support_vectors = svm.support_vectors_

print("Support Vectors:")
print(support_vectors)

# Q2d)
# Extract weights and bias
w = svm.weights
b = svm.bias
print("Vector w:", w)
print("Real number b:", b)
```
