# Perceptron Decision Boundaries

Perceptron training variants and decision-boundary visualization.

Stage: learning experiment.


```python
import numpy as np

class Perceptron:
    def __init__(self):
        self.weights = None
        self.bias = 0

    def initialize(self, num_features):
        # Initialize weights to zeros and bias to zero
        self.weights = np.zeros(num_features)
        self.bias = 0

    def predict(self, inputs):
        # Compute activation (weighted sum of inputs + bias)
        activation = np.dot(inputs, self.weights) + self.bias
        # Linear activation function
        return activation

    def train(self, X, y, epochs=100, learning_rate=0.1):
        num_features = X.shape[1]
        self.initialize(num_features)
        for _ in range(epochs):
            for inputs, label in zip(X, y):
                prediction = self.predict(inputs)
                error = label - prediction
                # Update weights and bias
                self.weights += learning_rate * error * inputs
                self.bias += learning_rate * error

# Example usage:
X_train = np.array([[2, 2], [1, -2], [-2, 2], [-1, 1]])
y_train = np.array([0, 1, 0, 1])
perceptron_model = Perceptron()
perceptron_model.train(X_train, y_train, epochs=100, learning_rate=0.6)
test_input = np.array([2, 2])
print("Prediction for test input:", perceptron_model.predict(test_input))
```

```python
import numpy as np
import matplotlib.pyplot as plt

class Perceptron:
    def __init__(self):
        self.weights = None
        self.bias = 0

    def initialize(self, num_features):
        # Initialize weights to zeros and bias to zero
        self.weights = np.zeros(num_features)
        self.bias = 0

    def predict(self, inputs):
        # Compute activation (weighted sum of inputs + bias)
        activation = np.dot(inputs, self.weights) + self.bias
        # Linear activation function
        return activation

    def train(self, X, y, epochs=100, learning_rate=0.1):
        num_features = X.shape[1]
        self.initialize(num_features)
        for _ in range(epochs):
            for inputs, label in zip(X, y):
                prediction = self.predict(inputs)
                error = label - prediction
                # Update weights and bias
                self.weights += learning_rate * error * inputs
                self.bias += learning_rate * error

    def plot_decision_boundary(self, X, y):
        # Plotting the decision boundary
        x1_min, x1_max = X[:, 0].min() - 1, X[:, 0].max() + 1
        x2_min, x2_max = X[:, 1].min() - 1, X[:, 1].max() + 1
        xx1, xx2 = np.meshgrid(np.arange(x1_min, x1_max, 0.1),
                               np.arange(x2_min, x2_max, 0.1))
        Z = self.predict(np.c_[xx1.ravel(), xx2.ravel()])
        Z = Z.reshape(xx1.shape)
        plt.contourf(xx1, xx2, Z, alpha=0.4)
        plt.xlim(xx1.min(), xx1.max())
        plt.ylim(xx2.min(), xx2.max())

        # Plotting data points
        for idx, cl in enumerate(np.unique(y)):
            plt.scatter(x=X[y == cl, 0], y=X[y == cl, 1],
                        alpha=0.8, c=[['blue', 'red'][idx]],
                        marker=['o', 'x'][idx], label=cl)

# Example usage:
X_train = np.array([[2, 2], [1, -2], [-2, 2], [-1, 1]])
y_train = np.array([0, 1, 0, 1])
perceptron_model = Perceptron()
perceptron_model.train(X_train, y_train, epochs=100, learning_rate=0.6)

plt.figure(figsize=(10, 6))
perceptron_model.plot_decision_boundary(X_train, y_train)
plt.title('Decision Boundary of the Perceptron')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend(loc='upper left')
plt.grid()
plt.show()
```

```python
import numpy as np
import math

class perc:

    def __init__(self):
        self.w = None
        self.b = 0
        self.af = None

    def initalize(self, n, activation_function):
        self.w = np.zeros(n)
        self.b = 0
        self.af = activation_function
        return

    def predict(self, inpu):
        a = np.dot(self.w, inpu) + self.b
        if self.af == "lin":
            return a
        elif self.af == "logsig":
            return 1 / (1 + np.exp(-a))
        elif self.af == "hardlim":
            return 1 if a > 0 else 0

    def train(self, x, y, epoch):
        self.initalize(x.shape[1], activation_function='hardlim')  # Setting the activation function here
        for _ in range(epoch):
            for inp, l in zip(x, y):
                e = l - self.predict(inp)
                self.w += inp * e
                self.b += e

p = np.array([[2, 2], [1, -2], [-2, 2],[-1 ,1]])
t = np.array([0, 1, 0,1])
a = perc()
a.train(p, t, 10)
result = a.predict(np.array([-1, 0]))
print(result)
```
