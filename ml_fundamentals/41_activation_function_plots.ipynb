# Activation Function Plots

Hard-limit, linear and sigmoid activation-function visualizations.

Stage: learning experiment.


```python
import numpy as np
import matplotlib.pyplot as plt

# Define the transfer functions
def hard_limit(x):
    return 1 if x > 0 else 0

def linear(x):
    return x

def log_sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Create input values for plotting
x_values = np.linspace(-2, 2, 100)

# Create plots for each transfer function
plt.figure(figsize=(15, 5))

# Plot for Hard limit transfer function
plt.subplot(1, 3, 1)
plt.plot(x_values, [hard_limit(x) for x in x_values], label='Hard Limit')
plt.xlabel('Input')
plt.ylabel('Output')
plt.title('Hard Limit Transfer Function')
plt.grid(True)
plt.legend()

# Plot for Linear transfer function
plt.subplot(1, 3, 2)
plt.plot(x_values, [linear(x) for x in x_values], label='Linear')
plt.xlabel('Input')
plt.ylabel('Output')
plt.title('Linear Transfer Function')
plt.grid(True)
plt.legend()

# Plot for Log-sigmoid transfer function
plt.subplot(1, 3, 3)
plt.plot(x_values, [log_sigmoid(x) for x in x_values], label='Log-sigmoid')
plt.xlabel('Input')
plt.ylabel('Output')
plt.title('Log-sigmoid Transfer Function')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
