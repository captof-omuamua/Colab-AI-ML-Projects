# Tamil Moderation Dataset Export

Load a Tamil hate-speech dataset and export its splits to CSV for moderation research.

Stage: data preparation.


```python
# Install the datasets library
!pip install -q datasets

from datasets import load_dataset
import pandas as pd

# Specify the directory where you want to store the dataset in Colab
cache_dir = "/content/synthetic_data"

# Load the dataset from Hugging Face and specify the cache directory
try:
    # Load each split separately
    dataset = load_dataset("krishan-CSE/Hate_Speech_Tamil", cache_dir=cache_dir)

    # Convert the 'train' and 'test' splits to Pandas DataFrames
    train_df = pd.DataFrame(dataset['train'])
    test_df = pd.DataFrame(dataset['test'])

    # Save the DataFrames as CSV files if needed
    train_df.to_csv(f"{cache_dir}/train.csv", index=False)
    test_df.to_csv(f"{cache_dir}/test.csv", index=False)

    print("Dataset loaded, stored, and saved as CSV files in specified directory.")
except Exception as e:
    print(f"An error occurred: {e}")
```
