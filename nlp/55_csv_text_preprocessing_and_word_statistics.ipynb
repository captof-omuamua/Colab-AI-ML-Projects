# Csv Text Preprocessing And Word Statistics

CSV text loading, punctuation cleanup, token counts and unigram probabilities.

Stage: learning experiment.


```python
import csv

def import_sentences_from_csv(file_path):
    sentences = []

    with open(file_path, 'r', encoding='utf-8') as csv_file:
        csv_reader = csv.reader(csv_file)

        # Assuming text data is in the first column (index 0)
        for row in csv_reader:
            sentences.append(row[0])

    return sentences

# Example usage:
csv_file_path = '/content/fake.csv'
sentences = import_sentences_from_csv(csv_file_path)

# Print the first few sentences as an example
print(sentences[:5])
```

```python
import re

def preprocess_text(sentences):
    cleaned_sentences = []

    for sentence in sentences:
        # Remove punctuation and convert to lowercase
        cleaned_sentence = re.sub(r'[^\w\s]', '', sentence.lower())
        # Remove extra whitespaces
        cleaned_sentence = ' '.join(cleaned_sentence.split())
        cleaned_sentences.append(cleaned_sentence)

    return cleaned_sentences

# Example usage:
preprocessed_sentences = preprocess_text(sentences)

# Print the preprocessed sentences as an example
print(preprocessed_sentences[:5])
```

```python
def tokenize_and_count(sentences):
    word_frequency = {}

    for sentence in sentences:
        # Split the sentence into words
        words = sentence.split()

        # Count the frequency of each word
        for word in words:
            if word in word_frequency:
                word_frequency[word] += 1
            else:
                word_frequency[word] = 1

    # Sort the word_frequency dictionary by frequency in descending order
    sorted_word_frequency = sorted(word_frequency.items(), key=lambda x: x[1], reverse=True)

    return sorted_word_frequency

# Example usage:
tokenized_and_counted = tokenize_and_count(preprocessed_sentences)

# Print the word and its frequency in descending order
for word, frequency in tokenized_and_counted:
    print(f"{word}: {frequency}")
```

```python
def calculate_word_probability(word, frequency, total_words):
    # Calculate the probability of the word
    probability = frequency / total_words
    return probability

def calculate_and_print_probabilities(tokenized_and_counted, total_words):
    # Calculate and print the probability of each word in descending order
    for word, frequency in tokenized_and_counted:
        probability = calculate_word_probability(word, frequency, total_words)
        print(f"{word}: {probability:.4f}")

# Calculate the total number of words
total_words = sum(frequency for _, frequency in tokenized_and_counted)

# Example usage:
calculate_and_print_probabilities(tokenized_and_counted, total_words)
```
