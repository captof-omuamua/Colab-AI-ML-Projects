#### 4. Write a python code to split the sentences into words

```python
import string

def split_text_into_words(text):
    # Split the text into sentences
    sentences = text.split('.')
    # Split each sentence into words
    words_in_sentences = []
    for sentence in sentences:
        # Remove leading and trailing whitespaces
        sentence = sentence.strip()
        # Split the sentence into words
        words = sentence.split()
        # Remove punctuation from words
        words = [word.translate(str.maketrans('', '', string.punctuation)) for word in words]
        # Remove empty words
        words = [word for word in words if word]
        words_in_sentences.append(words)
    return words_in_sentences

# Read the text from the file
with open('nlp_ass2.txt', 'r') as file:
    text = file.read()

# Split the text into words
words_in_sentences = split_text_into_words(text)

# Print the words in sentences
print("Words in Sentences:")
for i, words in enumerate(words_in_sentences, 1):
    print("Sentence", i, ":", words)
```

#### 6. Write a python code to split the words into subwords (split if no of characters are more than ‘k’ )

```python
def split_words_into_subwords(words, k):
    subwords_list = []
    for word in words:
        if len(word) > k:
            subwords = [word[i:i+k] for i in range(0, len(word), k)]
            subwords_list.extend(subwords)
        else:
            subwords_list.append(word)
    return subwords_list

k = 3
for word in words_in_sentences:
    subwords = split_words_into_subwords([word], k)  # Pass each word as a list
    print("Subwords:", subwords)
```

```python
def read_corpus(filename):
    with open(filename, 'r') as reader:
        lines = [line.rstrip('.\n') for line in reader]
    return lines

def tokenize_lines(lines):
    tokens = [line.split(' ') for line in lines]
    return [token for sublist in tokens for token in sublist]

def calculate_frequency(tokens):
    frequency_dict = {}
    for token in tokens:
        if token in frequency_dict:
            frequency_dict[token] += 1
        else:
            frequency_dict[token] = 1
    return frequency_dict

def calculate_probability(lines):
    tokens = tokenize_lines(lines)
    frequency_dict = calculate_frequency(tokens)
    probability_dict = {token: frequency_dict[token] / len(tokens) for token in frequency_dict}
    return probability_dict

# Read corpus from file
lines = read_corpus('nlp_ass2.txt')

# Calculate and display frequency of each token
frequency = calculate_frequency(tokenize_lines(lines))
print("Frequency of Tokens:", sorted(frequency.items(), key=lambda x: x[1], reverse=True), end="\n\n")

# Alternatively, you can calculate and display probability of each token
# probability = calculate_probability(lines)
# print("Probability of Tokens:", sorted(probability.items(), key=lambda x: x[1], reverse=True), end="\n\n")
```

#### 11. BAG OF WORDS



```python
import string

def read_sentences_from_file(filename):
    with open(filename, 'r') as file:
        text = file.read()
        # Split text into sentences (basic concept, no re library)
        sentences = text.split('.')
        # Tokenize each sentence
        tokenized_sentences = [tokenize(sentence) for sentence in sentences if sentence.strip()]
        return tokenized_sentences

# Function to tokenize a sentence
def tokenize(sentence):
    # Remove punctuation and convert to lowercase
    sentence = sentence.translate(str.maketrans('', '', string.punctuation)).lower()
    # Split the sentence into tokens
    tokens = sentence.split()
    return tokens

# Function to create a vocabulary from a list of tokenized sentences
def create_vocabulary(tokenized_sentences):
    vocabulary = set()
    for tokens in tokenized_sentences:
        vocabulary.update(tokens)
    return sorted(vocabulary)

# Function to represent a sentence using the Bag of Words method
def bag_of_words(sentence, vocabulary):
    sentence_representation = [0] * len(vocabulary)
    for word in sentence:
        if word in vocabulary:
            index = vocabulary.index(word)
            sentence_representation[index] += 1
    return sentence_representation

# Read text file and get all sentences
tokenized_sentences = read_sentences_from_file('nlp_ass2.txt')

# Create vocabulary
vocabulary = create_vocabulary(tokenized_sentences)
print("Vocabulary:", vocabulary)

# Represent each sentence using Bag of Words
for i, sentence in enumerate(tokenized_sentences, 1):
    representation = bag_of_words(sentence, vocabulary)
    print("Sentence", i, ":", representation)
```

#### 12. TERM FREQUENCY

```python
import random
import math

def calculate_term_frequency(sentence, all_words):
    words_in_document = sentence.split()
    tf = {}

    for word in all_words:
        count_word = words_in_document.count(word)
        if count_word != 0:
            tf[word] = math.log10(1 + count_word)  # Using 1 + count_word for smoothing
        else:
            tf[word] = 0

    return tf

# Generate a three-sentence document using the words randomly
def generate_document(words):
    document = ''
    for _ in range(6):
        document += words[random.randint(0, 11)] + ' '
    return document

# Term frequency
words = ['the','quick','brown','fox','jumps','over','the','lazy','dog','A','fast','brown','fox','jumps','over','a','lazy','dog']

# Generate three documents
document1 = generate_document(words)
document2 = generate_document(words)
document3 = generate_document(words)

print(f"document1 = {document1}")
print(f"document2 = {document2}")
print(f"document3 = {document3}")

# Calculate term frequency for each document
tf1 = calculate_term_frequency(document1, words)
tf2 = calculate_term_frequency(document2, words)
tf3 = calculate_term_frequency(document3, words)

print(f"document1 term frequency = {tf1}")
print(f"document2 term frequency = {tf2}")
print(f"document3 term frequency = {tf3}")
```

#### 13. INVERSE DOCUMENT FREQUENCY

```python
#idf

def calculate_inverse_document_frequency(all_documents, term):
    num_documents_with_term = sum(1 for document in all_documents if term in document)
    if num_documents_with_term != 0:
        return math.log10(len(all_documents) / num_documents_with_term)
    else:
        return 0

document1 = "The old oak tree stood tall and proud, its branches swaying gently in the breeze."
document2 = "Lost in thought, she wandered through the forest, the crunch of leaves beneath her feet."
document3 = "Determination gleamed in her eyes as she prepared for the challenge ahead."

all_documents = [document1, document2, document3]
print(all_documents)

# Create a set of unique words in all documents
words = set(word for document in all_documents for word in document.lower().split())

idf = {}
for word in words:
    idf[word] = calculate_inverse_document_frequency(all_documents, word)

print(idf)
```

### 1. Write a python code to find the probability of word sequence for a part-of-speech sequence.

```python
# Importing necessary libraries
import numpy as np
import pandas as pd
from itertools import combinations_with_replacement
import itertools
```

```python
# Function to calculate emission probabilities
def calculate_emission_probabilities(vocabulary, pos_tags, input_text):
    # Flatten the input text
    flattened_input = [item for sublist in input_text for item in sublist]

    # Initialize emission probability matrix with zeros
    emission_probabilities = np.zeros((len(vocabulary), len(pos_tags)))

    # Compute emission probabilities
    for i in range(len(vocabulary)):
        for j in range(len(pos_tags)):
            for k in flattened_input:
                if k[0] == vocabulary[i] and k[1] == pos_tags[j]:
                    emission_probabilities[i][j] += 1

    # Convert ndarray to a DataFrame
    emission_probabilities_df = pd.DataFrame(emission_probabilities, columns=pos_tags)

    # Insert a column at the beginning to represent the words
    emission_probabilities_df.insert(0, "Words", vocabulary)

    # Divide the values in each tag column by that column sum to get the emission probabilities
    for col in pos_tags:
        emission_probabilities_df[col] = emission_probabilities_df[col] / emission_probabilities_df[col].sum()

    # Reset the index to the words/vocabulary instead of index numbers
    emission_probabilities_df.set_index('Words', inplace=True)

    return emission_probabilities_df

# Function to calculate transition probabilities
def calculate_transition_probabilities(vocabulary, pos_tags, input_text):
    # Modify tags to include start and end of sentence tags
    modified_tags = pos_tags.copy()
    modified_tags.insert(0, "<s>")
    modified_tags.append("</s>")

    # Suppress 0/0 warning
    np.seterr(divide='ignore', invalid='ignore')

    # Initialize transition probability matrix with zeros
    transition_probabilities = np.zeros((len(modified_tags), len(modified_tags)))

    # Compute transition probabilities
    for sentence in input_text:
        for index in range(len(sentence) - 1):
            if index == 0:
                transition_probabilities[modified_tags.index("<s>")][modified_tags.index(sentence[index][1])] += 1
            transition_probabilities[modified_tags.index(sentence[index][1])][modified_tags.index(sentence[index+1][1])] += 1
        transition_probabilities[modified_tags.index(sentence[len(sentence)-1][1])][modified_tags.index("</s>")] += 1

    # Normalize transition probabilities
    transition_probabilities = transition_probabilities / transition_probabilities.sum(axis=1, keepdims=True)

    # Replace nan values with 0
    transition_probabilities = np.nan_to_num(transition_probabilities)

    # Convert ndarray to DataFrame
    transition_probabilities_df = pd.DataFrame(transition_probabilities, columns=modified_tags)

    # Insert tags column with the modified tags list and set it as index
    transition_probabilities_df.insert(0, "Tags", modified_tags)
    transition_probabilities_df.set_index("Tags", inplace=True)

    return transition_probabilities_df
```

```python
def calculate_probability_word_given_POS(word_sequence, pos_sequence, emission_probabilities, transition_probabilities):
    # Splitting the words and tags from the word and pos sequences
    words = word_sequence.split()
    pos = pos_sequence.split()

    # Creating an array to store the state probabilities for each pos state given
    probs = np.zeros(len(words))

    # Initialize total probability
    total_probability = transition_probabilities[pos[0]]["<s>"]

    for i in range(len(words)):
        # Storing the emission probability first
        probs[i] = emission_probabilities[pos[i]][words[i]]

        # Multiplying with the transition probability
        # If we are in the last word of the sentence, we need to multiply the transition probability of that word to
        # the end of sentence tag
        if i != len(words) - 1:
            probs[i] *= transition_probabilities[pos[i+1]][pos[i]]
        else:
            probs[i] *= transition_probabilities["</s>"][pos[i]]

    # Returning the product of the all the state probabilities and multiplying it with the start probability
    return total_probability * np.prod(probs)
```

```python
# Sample sentences given to find the transition and emission probabilities
sample_sentences = [
    "It/PRP looks/VB like/IN circle/N",
    "He/PRP like/IN chocolate/N",
    "I/PRP like/VB chocolate/N"
]

# Convert each sentence into a list of word-tag pairs
input_data = []
for sentence in sample_sentences:
    words_tags_pairs = []
    for word_tag in sentence.split():
        word, tag = word_tag.split('/')
        words_tags_pairs.append((word, tag))
    input_data.append(words_tags_pairs)

input_data
```

```python
# Get a list of unique tags and words
unique_tags = set()
unique_vocab = set()

for sentence in input_data:
    for word_tag_pair in sentence:
        unique_tags.add(word_tag_pair[1])
        unique_vocab.add(word_tag_pair[0])

unique_tags = list(unique_tags)
unique_vocab = list(unique_vocab)

print("Unique tags:", unique_tags)
print("Unique vocabulary:", unique_vocab)
```

```python
# Calling and storing the emission probability DataFrame
emission_probabilities_df = calculate_emission_probabilities(unique_vocab, unique_tags, input_data)
emission_probabilities_df
```

```python
# Calling and storing the transition probability DataFrame
transition_probabilities_df = calculate_transition_probabilities(unique_vocab, unique_tags, input_data)
transition_probabilities_df
```

### 3. Implement Viterbi to find the maximum probable POS tag sequence


```python
# training tagged data
train_data = [item for sublist in input_data for item in sublist]
train_data
```

```python
def viterbi_algorithm(words, train_data):
    # words is the untagged word sequence we want to find the POS sequence of
    state_sequence = []

    # Get the set of all POS tags from the training data
    tags = list(set([pair[1] for pair in train_data]))

    # Iterate over each word in the input sequence
    for key, word in enumerate(words):
        # Initialize a list to store the probabilities for each tag
        probabilities = []

        # For each tag in the tag list
        for tag in tags:
            # If key is 0, we are in the first word of the word sequence
            # So, we need to get transition probability from start tag to the first word of the sentence
            if key == 0:
                transition_probability = transition_probabilities_df[tag]["<s>"]
            else:
                # Compute transition probabilities based on the previous state to the current state
                transition_probability = transition_probabilities_df[tag][state_sequence[-1]]

            # Emission probability
            emission_probability = emission_probabilities_df[tag][word]

            # State probability
            state_probability = emission_probability * transition_probability

            # Append the state probability
            probabilities.append(state_probability)

        # Get the maximum state probability and find out which state the probability belongs to
        max_probability = max(probabilities)
        # Get the state for which probability is maximum
        max_state = tags[probabilities.index(max_probability)]

        # Append the state with max probability
        state_sequence.append(max_state)

    # Return the respective word and its state/tag as a list of tuples
    return list(zip(words, state_sequence))

word_seq = "He looks like circle"
print("Maximum probable POS tag sequence for the word sequence", word_seq, "is : ")
result_sequence = viterbi_algorithm(word_seq.split(), train_data)
print(result_sequence)

```
