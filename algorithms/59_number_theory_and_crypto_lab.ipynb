polymult

```python
def polymult(x,y):
    result = []
    for i in range(len(x)):
        for j in range(len(y)):
                        temp=x[i]*y[j]
                        result.append(temp)
    return result
```

decimal and binary 

```python
def dec2bin(decimal):
    decimal = int(decimal)
    if decimal==0:
            return "0000"
    binary= ""
    while decimal>0:
            binary+=str(decimal%2)
            decimal = decimal // 2
    binary=binary[::-1]
    return  str(binary)

def bin2dec(binary):
    binary = str(binary)
    decimal = 0
    for i in binary:
        decimal = decimal * 2 + int(i)

    return str(decimal)

def dec2hex(n):
        n = int(n)
        if n == 0:
                return 0
        digits = '0123456789ABCDEF'
        hex = ''

        while n>0:
                rem = n%16 #remainder 
                hex = digits[rem] + hex
                n//=16 # quitent 
        return str(hex)

# hexa to decimal

def hex2dec(hex):
        digits = '0123456789ABCDEF'
        n = 0
        for i,digit in enumerate(hex[::-1]):
                n+= digits.index(digit)*(16**i)
        return str(n)


def hex2bin(hex):

    bin = hex2dec(hex)
    bin = dec2bin(bin)

    if len(bin)!= len(hex)*4:
           bin = '0'+str(bin)
    return str(bin)
    
    
def bin2hex(bin):
    
    hex = bin2dec(bin)
    hex = dec2hex(hex)

    return str(hex)

```

gcd

```python
def gcd(x,y):
        if (x==0 and y==0):
                return "Invalid or 0 "
        elif (y==0):
                return x
        else:
                return gcd(y, x%y)


#----------------------------- GCD list -------------------------------#

def gcdlist(mylist):
    if len(mylist)==1:
        return mylist[0]
    return gcd(mylist[0],gcdlist(mylist[1:]))

    
#-----------------------------Extended GCD-------------------------------#
def egcd(a,b):
        # return(gcd, coeff_A, coeff_B)
        if a ==0:
                return b,0,1
        gcd ,x1,y1 = egcd(b%a,a)

        x= y1- (b//a)*x1
        y= x1
        return gcd ,x,y
```

```python

def modinverse(a,p):
    # return mod inverse of a under (mod p)
    gd,x,y=egcd(a,p)
    if gd!=1:
        return -1
    return (x%p+p)%p

def fastmodexp(c,x,m):
    return pow(c,x,m)
```

```python
###### odd magic sqare ###########

def oddmagicsq(n):    
    ms= np.zeros((n,n))

    if n%2==0:
        return -1

    num=1
    row,col=0,n//2
    while num <=  n**2:
        ms[row][col]=num
        num+=1
        new_row,new_col=(row-1)%n,(col+1)%n
        if ms[new_row][new_col]==0:
            row,col=new_row,new_col
        else:
            row=(row+1)%n

    return ms
```

```python
def DES_hex2bin(s):
    hb={'0':'0000','1':'0001','2':'0010','3':'0011','4':'0100',
        '5':'0101','6':'0110','7':'0111','8':'1000','9':'1001',
        'a':'1010','A':'1010','b':'1011','B':'1011','c':'1100','C':'1100',
        'd':'1101','D':'1101','e':'1110','E':'1110','f':'1111','F':'1111'}
    n=len(s)
    s1=''
    for i in range(int(n)):
        s1+=hb.get(s[i])
    return s1
#------------------------------------------------------------------
def DES_bin2hex(s):
    bh={'0000': '0','0001': '1','0010': '2','0011': '3','0100': '4','0101': '5',
     '0110': '6','0111': '7','1000': '8','1001': '9','1010': 'A','1011': 'B',
     '1100': 'C','1101': 'D','1110': 'E','1111': 'F'}
    n=len(s)/4
    s1=''
    for i in range(int(n)):
        s1 += bh.get(s[i*4:(i+1)*4])
    return s1


def DES_ip(inBitArray):  # using list
    outBitArray = ''
    for i in IP:
        outBitArray += inBitArray[i - 1]
    return outBitArray
#-------------------------------------------------------------------
def DES_invip(inBitArray):
    outBitArray = ''
    for i in range(64):
        outBitArray += inBitArray[IP.index(i + 1)]        
    return outBitArray
#-------------------------------------------------------------------
def DES_NLeftShift(Iter):
    if(Iter == 1 or Iter == 2 or Iter == 9 or Iter == 16):
        return 1
    elif((Iter >= 3 and Iter <= 8) or (Iter >= 10 and Iter <= 15)):
        return 2
#-------------------------------------------------------------------
def leftshift(s1,i):
    s2=s1[i:]+s1[0:i]
    return s2
#-------------------------------------------------------------------    
def DES_pc1(keyString):
    outBitArray=''
    for i in range(56):
        outBitArray += keyString[PC1[i] - 1]        
    return outBitArray
#-------------------------------------------------------------------
def DES_pc2(keyString):
    outBitArray=''
    for i in range(48):
        outBitArray += keyString[PC2[i] - 1]        
    return outBitArray
#-------------------------------------------------------------------
def DES_e(keySubString):
    outBitArray=''
    for i in range(48):
        outBitArray += keySubString[E[i] - 1]        
    return outBitArray
#-------------------------------------------------------------------
def DES_p(keySubString):
    outBitArray=''
    for i in range(32):
        outBitArray += keySubString[P[i] - 1]        
    return outBitArray
#-------------------------------------------------------------------
def DES_s(keySubString):
    outBitArray = ''
    for i in range(8):
        # b4array = list('0000')
        rk6 = keySubString[i*6:(i+1)*6]
        ri = int(rk6[0]+rk6[5],2)   # row index (0 - 3)
        ci = int(rk6[1:5],2)        # col index (0 - 15) 
        di=str(bin(S[i][ri][ci])[2:])

        if(len(di)==1):
            di='000'+di
        elif(len(di)==2):
            di='00'+di
        elif(len(di)==3):
            di='0'+di
        outBitArray += di
    return outBitArray
#------------------------------------------------------------------
def DES_xor(s1,s2):
    if(len(s1)==len(s2)):
        s3=''
        for i in range(len(s1)):
            s3 += str(int(s1[i]) ^ int(s2[i]))
        return s3
    else:
        return 'Err: len(s1) .NE. len(s2)'
#------------------------------------------------------------------
def DES_f(r,k):
    r1 = DES_e(r)
    s1 = DES_xor(r1,DES_hex2bin(k))
    s2 = DES_s(s1)
    s3 = DES_p(s2)
    return s3

#------------------------------------------------------------------
def DES_subkeys(k):
    k1=DES_hex2bin(k)
    k2=DES_pc1(k1)
    ki=[]
    for i in range(16):
        C,D=k2[0:28],k2[28:56]
        n=DES_NLeftShift(i+1)
        C1,D1=leftshift(C, n),leftshift(D, n)
        ki.append(DES_bin2hex(DES_pc2(C1+D1)))
        k2=C1+D1
    return ki     
#------------------------------------------------------------------
def DES_encrypt(pt,k): # pt & k in hex
    pt1=DES_hex2bin(pt)
    d1=DES_ip(pt1)
    Lo , Ro = d1[0 : 32] , d1[32 : 64]
    ki=DES_subkeys(k)
    for i in range(16):
        L1 = Ro
        R1 = DES_xor(Lo,DES_f(Ro,ki[i]))
        Lo,Ro = L1,R1    
    po = Ro + Lo
    d2 = DES_invip(po)
    ct = DES_bin2hex(d2)
    return ct
#------------------------------------------------------------------
def DES_decrypt(ct,k):  
    d3 = DES_ip(DES_hex2bin(ct)) 
    Ro,Lo = d3[0:32],d3[32:64]
    ki = DES_subkeys(k)
    for i in range(16):
        R1 = Lo
        L1 = DES_xor(Ro,DES_f(R1, ki[15-i]))
        Lo = L1
        Ro = R1
    d3 = Lo + Ro
    d4 = DES_invip(d3)
    d5 = DES_bin2hex(d4)
    return d5
#------------------------------------------------------------------
def DES3_encrypt(pt1,k1,k2,k3):
    ct1=DES_encrypt(pt1,k1)
    ct2=DES_decrypt(ct1,k2)
    ct3=DES_encrypt(ct2,k3)
    return ct3
#------------------------------------------------------------------
def DES3_decrypt(ct1,k1,k2,k3):
    pt3=DES_decrypt(ct1,k3)
    pt2=DES_encrypt(pt3,k2)
    pt1=DES_decrypt(pt2,k1)
    return pt1
#------------------------------------------------------------------
```

```python
########################### RSA ###########################
def RSA_genkey(p,q):                    
    # return a list of 2 tuple [(n,e),(n,d)].
    # public key (n,e) and private key (n,d)
    #pand q are primes
    n=p*q
    phi=(p-1)*(q-1)
    e=2
    while gcd(e,phi)!=1:
        e+=1
    d=modinverse(e,phi)
    return [(n,e),(n,d)]

def fastmodexp(a,n,p):
    # a ^ n (mod p).
    if n==0:
        return 1
    if n%2==0:
        return (fastmodexp(a,n//2,p)**2)%p
    return (a*fastmodexp(a,n-1,p))%p
def RSA_sign(digest,priv):              
    # (digest ^ d) % n
    return fastmodexp(digest,priv[1],priv[0])
    
def RSA_decrypt(ekey,priv):             # (ekey ^ d) % n
    return fastmodexp(ekey,priv[1],priv[0])

def RSA_verify(digest,pub):          # (signature ^ e) % n
    return fastmodexp(digest,pub[1],pub[0])

def RSA_encrypt(key,pub):               # (key ^ e) % n
    return fastmodexp(key,pub[1],pub[0])
    
```

```python
#######################  ECC  ########################

# assume the point at infinity, (additive identity) as 'Inf'
def ECC_valid(a,b,p):                  
    # Check if the given elliptic curve is singular or not, if p is a prime 
    # equivalent to 3(mod 4)
    if (4*(a**3)+27*(b**2))%p==0:
        return f"the given {a,b,p} elliptic curve is singular"
    return f"the given {a,b,p} elliptic curve is not singular"


def ECC_genpts(a,b,p):                  
    # generate and return a list of all points in the modular elliptic curve, as a list 
    # assume the point at infinity, (additive identity) as 'Inf'
    points=[]
    for x in range(p):
        for y in range(p):
            if (y**2)%p==((x**3)+(a*x)+b)%p:
                points.append((x,y))
    return points


def ECC_plotPxy(a,b,p):
    # plot the points in the modular elliptic curve
    points=ECC_genpts(a,b,p)
    x=[]
    y=[]
    for i in points:
        x.append(i[0])
        y.append(i[1])
    plt.scatter(x,y)
    plt.show()


def ECC_orderP(P,a,b,p):
    # given P(x,y) in E, find the order of P
    points=ECC_genpts(a,b,p)
    if P not in points:
        return "P is not in E"
    count=0
    while P!="Inf":
        count+=1
        P=ECC_addPQ(P,P,a,b,p)
    return count


def ECC_addPQ(P,Q,a,b,p):
    # return the coordinates of P(x1,y1) + Q(x2,y2)
    if P=="Inf":
        return Q
    if Q=="Inf":
        return P
    if P==Q:
        if P[1]==0:
            return "Inf"
        s=((3*(P[0]**2)+a)*modinverse(2*P[1],p))%p
    else:
        if P[0]==Q[0]:
            return "Inf"
        s=((P[1]-Q[1])*modinverse(P[0]-Q[0],p))%p
    x=(s**2-P[0]-Q[0])%p
    y=(s*(P[0]-x)-P[1])%p
    return (x,y)


def ECC_isPinE(P,a,b,p):
    # return True if P belongs to E, or False
    points=ECC_genpts(a,b,p)
    if P in points:
        return True
    return False


def ECC_mul(P,n,a,b,p):
	# return the coordinates of nP
	if n==0:
		return "Inf"
	if n==1:
		return P
	if n%2==0:
		return ECC_addPQ(ECC_mul(P,n//2,a,b,p),ECC_mul(P,n//2,a,b,p),a,b,p)
	return ECC_addPQ(P,ECC_mul(P,n-1,a,b,p),a,b,p)


def ECC_nP(n,P,a,b,p):
	# return the coordinates of P + ... + P, i.e., P added n times to itself
	return ECC_mul(P,n,a,b,p)

    
def ECC_2nP(P,a,b,p):
	# return the coordinates of 2np
	n=2
	# while ECC_mul(P,n,a,b,p)!="Inf":
	# 	n*=2
	return ECC_mul(P,n,a,b,p)
    

```
