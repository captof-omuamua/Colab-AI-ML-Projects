```python
X = librosa.stft(x, n_fft = 512, win_length=400, window = 'hann',)
Xdb = librosa.amplitude_to_db(abs(X))
plt.figure(figsize=(10,5))
librosa.display.specshow(Xdb, sr=sr, x_axis='time', y_axis='hz')
plt.colorbar()
```

```python
X = librosa.stft(x, n_fft = 256, win_length=128, window = 'hann',)
Xdb = librosa.amplitude_to_db(abs(X))
plt.figure(figsize=(10,5))
librosa.display.specshow(Xdb, sr=sr, x_axis='time', y_axis='hz')
plt.colorbar()
```

```python
sig=x[:5000]
plt.figure(figsize=(10, 6))
plt.plot(sig)
plt.grid()
```

```python
zero_crossings = librosa.zero_crossings(sig, pad=False)
print(sum(zero_crossings))
```

```python
#spectral_centroids = librosa.feature.spectral_centroid(x, sr=sr)
cent = librosa.feature.spectral_centroid(y=x, sr=sr)
cent.shape
```

```python
S, phase = librosa.magphase(librosa.stft(y=x))
times = librosa.times_like(cent)
fig, ax = plt.subplots()
librosa.display.specshow(librosa.amplitude_to_db(S, ref=np.max),

                         y_axis='log', x_axis='time', ax=ax)

ax.plot(times, cent.T, label='Spectral centroid', color='w')
ax.legend(loc='upper right')
ax.set(title='log Power spectrogram')
```

```python
y=x[8001:8500]
plt.plot(y)
```

```python
acf=librosa.autocorrelate(y)
plt.plot(acf)
```

```python
a = librosa.lpc(y, order=14)
b = np.hstack([[0], -1 * a[1:]])
y_hat = scipy.signal.lfilter(b, [1], y)

fig, ax = plt.subplots()
ax.plot(y_hat)
ax.set_title('LP Model Forward Prediction')
```

```python
er=y-y_hat
plt.plot(er)
```
