# Discrete Fourier Transform

Explicit DFT implementation and short-time analysis of supplied audio.

Stage: learning experiment.

3rd question

```python
import librosa
import matplotlib.pyplot as plt
import numpy as np
signal, sampling_rate =librosa.load('/content/raw.wav')


window_size = int(0.03 * sampling_rate)

start_sample = 15000
frame_signal = signal[start_sample:start_sample + window_size]

N = 512

def compute_dft(signal):
    N = len(signal)
    dft_frame = np.zeros(N, dtype=complex)
    for k in range(N):
        for n in range(N):
            dft_frame[k] += signal[n] * np.exp(-2j * np.pi * k * n / N)
    return dft_frame


dft_frame = compute_dft(frame_signal)

frequency_axis = np.arange(N) * sampling_rate / N

plt.figure(figsize=(10, 6))
plt.plot(frequency_axis[:N//2], np.abs(dft_frame)[:N//2])
plt.title('Short-Time Fourier Transform')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude')
plt.grid(True)
plt.show()

# Find the index of the peak magnitude
peak_index = np.argmax(np.abs(dft_frame)[:N//2])

# Get the corresponding frequency
fundamental_frequency = frequency_axis[peak_index]

print("Fundamental Frequency:", fundamental_frequency, "Hz")
```
