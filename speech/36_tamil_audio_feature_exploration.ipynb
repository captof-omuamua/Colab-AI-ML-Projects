```python
def feature_chromagram(waveform, sample_rate):
  stft_spectrogram = np.abs(librosa.stft(waveform, n_fft=512, hop_length=200, win_length=400, window="hann"))
  chromagram = np.mean(librosa.feature.chroma_stft(S=stft_spectrogram, sr=sample_rate).T, axis=0)
  return chromagram

def feature_melspectrogram(waveform, sample_rate):
  melspectrogram = np.mean(librosa.feature.melspectrogram(y=waveform, sr=sample_rate, n_mels=128, fmax=8000, n_fft=512, hop_length=288, win_length=400, window="hann").T, axis=0)
  return melspectrogram

def feature_mfcc(waveform, sample_rate):
  mfc_coefficients = np.mean(librosa.feature.mfcc(y=waveform, sr=sample_rate, n_mfcc=40).T, axis=0)
  return mfc_coefficients
```

```python
def get_features(file):

   # Load the audio waveform and get the sampling rate
  waveform, sample_rate = librosa.load(file, sr=16000)

  # Compute features using functions defined above
  chromagram = feature_chromagram(waveform, sample_rate)
  melspectrogram = feature_melspectrogram(waveform, sample_rate)
  mfc_coefficients = feature_mfcc(waveform, sample_rate)

  # Create a feature matrix by horizontally stacking the feature vectors
  feature_matrix = np.hstack((chromagram, melspectrogram, mfc_coefficients))

  return feature_matrix
```

```python
def get_feature_matrix(files, labels):
  temp_feats = list()
  temp_labels = list()

  # Iterate over the files and labels
  for file, label in zip(files, labels):
    # Get features for each audio file
    features = get_features(file)  # Calls the previously defined function
    temp_feats.append(features)
    temp_labels.append(label)

  return np.array(temp_feats),np.array(temp_labels)
```

```python
train_features, train_labels = get_feature_matrix(train_files, train_labels)
test_features, test_labels = get_feature_matrix(test_files, test_labels)

train_features.shape
```

```python
#Check chromagram feature values
def get_features_statistics(df):

  features_df_chromagram = df.loc[:, :11]  # assuming the first 11 columns are chromagram features

  # Min and max values of each chromagram feature
  chroma_min = features_df_chromagram.min().min()
  chroma_max = features_df_chromagram.max().max()

  # mean of all chromagram features (after stacking them into a single series)
  chroma_mean = features_df_chromagram.stack().mean()

  # standard deviation of all chromagram features (after stacking them into a single series)
  chroma_stdev = features_df_chromagram.stack().std()

  # Print the results
  print(f'Chromagram features: \nmin: {chroma_min:.3f}, \nmax: {chroma_max:.3f}, \nmean: {chroma_mean:.3f}, \ndeviation: {chroma_stdev:.3f}')

  #Check mel spectrogram feature values
  features_df_melspectrogram = df.loc[:, 12:139]  # assuming columns 12 to 139 are mel spectrogram features

  mel_min = features_df_melspectrogram.min().min()
  mel_max = features_df_melspectrogram.max().max()

  # mean of all mel spectrogram features (after stacking them into a single series)
  mel_mean = features_df_melspectrogram.stack().mean()

  # standard deviation of all mel spectrogram features (after stacking them into a single series)
  mel_stdev = features_df_melspectrogram.stack().std()

  # Print the results
  print(f'\nMel Spectrogram features: \nmin: {mel_min:.3f}, \nmax: {mel_max:.3f}, \nmean: {mel_mean:.3f}, \ndeviation: {mel_stdev:.3f}')

  #Check MFCC feature values
  features_df_mfcc = df.loc[:,140:179]

  mfcc_min = features_df_mfcc.min().min()
  mfcc_max = features_df_mfcc.max().max()

  #stack all features into a single series so we don't get a mean of means or stdev of stdevs
  mfcc_mean = features_df_mfcc.stack().mean()

  mfcc_stdev = features_df_mfcc.stack().std()

  #print the results
  print(f' \n MFCC features:\
  min = {mfcc_min:.3f},\
  max = {mfcc_max:.3f};\
  mean = {mfcc_mean:.3f},\
  deviation={mfcc_stdev:.3f}')

```
