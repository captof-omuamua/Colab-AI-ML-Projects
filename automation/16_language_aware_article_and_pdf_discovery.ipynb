# Language Aware Article And Pdf Discovery

HTML article extraction, language-based text folders and PDF-link discovery.

Stage: prototype.

```python
import requests
from bs4 import BeautifulSoup
import os
from langdetect import detect

def scrape_newspapers(url):
    # Fetch the webpage
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    # Find all newspaper articles (you'll need to adjust this based on the website's structure)
    articles = soup.find_all('div', class_='article')

    for article in articles:
        # Extract title and content (adjust as needed)
        title = article.find('h2').text
        content = article.find('div', class_='content').text

        # Detect language
        language = detect(content)

        # Create language folder if it doesn't exist
        if not os.path.exists(language):
            os.makedirs(language)

        # Save article to appropriate folder
        with open(f"{language}/{title}.txt", 'w', encoding='utf-8') as f:
            f.write(content)

# Use the function
scrape_newspapers('https://www.careerswave.in/newspaper-download-pdf-free/')
```

```python
import requests
from bs4 import BeautifulSoup

# URL of the website
url = "https://www.careerswave.in/newspaper-download-pdf-free/"

# Custom headers to mimic a browser request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8,application/signed-exchange;v=b3;q=0.9",
    "Accept-Encoding": "gzip, deflate, br",
    "Accept-Language": "en-US,en;q=0.9",
    "Connection": "keep-alive"
}

# Send a GET request to fetch the raw HTML content
response = requests.get(url, headers=headers)

# Check if the request was successful
if response.status_code == 200:
    # Parse the content with BeautifulSoup
    soup = BeautifulSoup(response.content, "html.parser")

    # Find all links to PDF files
    pdf_links = []
    for link in soup.find_all('a', href=True):
        if link['href'].endswith('.pdf'):
            pdf_links.append(link['href'])

    # Print the PDF links
    for pdf in pdf_links:
        print(pdf)
else:
    print("Failed to retrieve the webpage. Status code:", response.status_code)
```

```python
import requests
from bs4 import BeautifulSoup

# URL of the website
url = "https://www.careerswave.in/newspaper-download-pdf-free/"

# Custom headers to mimic a browser request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8,application/signed-exchange;v=b3;q=0.9",
    "Accept-Encoding": "gzip, deflate, br",
    "Accept-Language": "en-US,en;q=0.9",
    "Connection": "keep-alive"
}

# Send a GET request to fetch the raw HTML content
response = requests.get(url, headers=headers)

# Check if the request was successful
if response.status_code == 200:
    # Set encoding to avoid decoding issues
    response.encoding = response.apparent_encoding

    # Parse the content with BeautifulSoup
    soup = BeautifulSoup(response.text, "lxml")

    # Find all links to PDF files
    pdf_links = []
    for link in soup.find_all('a', href=True):
        if link['href'].endswith('.pdf'):
            pdf_links.append(link['href'])

    # Print the PDF links
    for pdf in pdf_links:
        print(pdf)
else:
    print("Failed to retrieve the webpage. Status code:", response.status_code)
```
